# 01. Power Grid Transformer Telemetry - Exploratory Data Analysis (EDA)
**Power Grid AI - Predictive Maintenance & Grid Reliability**

### Objectives:
- Ingest raw SCADA electrical and chemical telemetry from substation transformers.
- Analyze distributions of thermal parameters (top-oil, winding), electrical load, and mechanical vibration.
- Investigate Dissolved Gas Analysis (DGA) concentrations ($H_2, CH_4, C_2H_2$).
- Understand correlations between load factors and hot-spot temperature rise.


```
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
%matplotlib inline

# Load raw transformer telemetry
raw_data_path = os.path.join('..', 'data', 'raw', 'transformer_data.csv')
df = pd.read_csv(raw_data_path)
print(f"Loaded dataset: {df.shape[0]} rows, {df.shape[1]} columns")
df.head()
```

### Telemetry Overview & Summary Statistics

```
df.info()
display(df.describe().round(2))
```

### Thermal & Electrical Distributions

```
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

sns.histplot(df['load_percentage'], kde=True, ax=axes[0, 0], color='royalblue')
axes[0, 0].set_title('Distribution of Transformer Load (%)', fontsize=12, fontweight='bold')

sns.histplot(df['winding_temperature_c'], kde=True, ax=axes[0, 1], color='crimson')
axes[0, 1].set_title('Distribution of Winding Temperature (°C)', fontsize=12, fontweight='bold')

sns.histplot(df['oil_temperature_c'], kde=True, ax=axes[1, 0], color='darkorange')
axes[1, 0].set_title('Distribution of Top Oil Temperature (°C)', fontsize=12, fontweight='bold')

sns.histplot(df['vibration_mm_s'], kde=True, ax=axes[1, 1], color='purple')
axes[1, 1].set_title('Distribution of Mechanical Vibration (mm/s)', fontsize=12, fontweight='bold')

plt.tight_layout()
plt.show()
```

### Correlation Matrix of Operating Parameters

```
numeric_cols = [
    'voltage_kV', 'current_A', 'load_percentage', 'oil_temperature_c',
    'winding_temperature_c', 'ambient_temperature_c', 'vibration_mm_s',
    'dissolved_hydrogen_ppm', 'dissolved_methane_ppm', 'dissolved_acetylene_ppm', 'stress_index'
]

plt.figure(figsize=(11, 8))
corr = df[numeric_cols].corr()
sns.heatmap(corr, annot=True, fmt=".2f", cmap='coolwarm', cbar=True)
plt.title('Sensor Telemetry Correlation Matrix', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

### Dissolved Gas Concentrations (DGA) across Assets

```
fig, axes = plt.subplots(1, 3, figsize=(16, 5))

sns.boxplot(data=df, x='transformer_id', y='dissolved_hydrogen_ppm', ax=axes[0], palette='Blues_d')
axes[0].set_title('Hydrogen (H2) by Asset', fontweight='bold')

sns.boxplot(data=df, x='transformer_id', y='dissolved_methane_ppm', ax=axes[1], palette='Greens_d')
axes[1].set_title('Methane (CH4) by Asset', fontweight='bold')

sns.boxplot(data=df, x='transformer_id', y='dissolved_acetylene_ppm', ax=axes[2], palette='Reds_d')
axes[2].set_title('Acetylene (C2H2) by Asset', fontweight='bold')

plt.tight_layout()
plt.show()
```
